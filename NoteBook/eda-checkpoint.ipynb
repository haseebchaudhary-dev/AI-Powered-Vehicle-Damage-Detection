```python
DATASET_ROOT = Path("dataset/CarDD_COCO")

AUDIT_DIR = Path("Results/EDA")

AUDIT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_SEED = 42
random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

print(f"Dataset root : {DATASET_ROOT}")
print(f"Audit folder : {AUDIT_DIR}")

if not DATASET_ROOT.exists():
    raise FileNotFoundError(
        f"\nDataset not found:\n{DATASET_ROOT}\n\n"
        "Please update DATASET_ROOT in CELL 02."
    )

print("\nDataset root exists.")
```

```python
max_depth = 4

for root, dirs, files in os.walk(DATASET_ROOT):
    root_path = Path(root)
    relative = root_path.relative_to(DATASET_ROOT)

    depth = len(relative.parts)

    if depth > max_depth:
        dirs[:] = []
        continue

    indent = "    " * depth
    folder_name = root_path.name

    print(f"{indent}{folder_name}/")

    for file in sorted(files)[:20]:
        print(f"{indent}    ├── {file}")

    if len(files) > 20:
        print(f"{indent}    └── ... ({len(files) - 20} more files)")
```

```python
json_files = list(DATASET_ROOT.rglob("*.json"))

if not json_files:
    raise FileNotFoundError(
        "No JSON files were found inside the dataset."
    )

for idx, path in enumerate(json_files, start=1):
    size_mb = path.stat().st_size / (1024 ** 2)

    print(
        f"{idx:02d}. {path.relative_to(DATASET_ROOT)} "
        f"({size_mb:.2f} MB)"
    )
```

```python
coco_candidates = []

for json_path in json_files:

    try:
        with open(json_path, "r", encoding="utf-8") as f:
            data = json.load(f)

        required_keys = {"images", "annotations", "categories"}

        if required_keys.issubset(data.keys()):
            coco_candidates.append(
                {
                    "path": json_path,
                    "data": data
                }
            )

    except Exception as e:
        print(
            f"Could not parse {json_path.name}: {e}"
        )


if not coco_candidates:
    raise ValueError(
        "No valid COCO annotation JSON was detected."
    )

for idx, candidate in enumerate(coco_candidates, start=1):

    data = candidate["data"]

    print(
        f"{idx}. {candidate['path'].relative_to(DATASET_ROOT)}"
    )

    print(
        f"   Images      : {len(data['images']):,}"
    )

    print(
        f"   Annotations : {len(data['annotations']):,}"
    )

    print(
        f"   Categories  : {len(data['categories']):,}"
    )
```

```python
selected_coco = max(
    coco_candidates,
    key=lambda x: len(x["data"]["images"])
)

ANNOTATION_FILE = selected_coco["path"]
COCO = selected_coco["data"]

print(
    "Annotation file:",
    ANNOTATION_FILE.relative_to(DATASET_ROOT)
)

print(
    "Images:",
    f"{len(COCO['images']):,}"
)

print(
    "Annotations:",
    f"{len(COCO['annotations']):,}"
)

print(
    "Categories:",
    f"{len(COCO['categories']):,}"
)
```

```python
print("VALIDATING IMAGE FILES")

all_files = [
    p for p in DATASET_ROOT.rglob("*")
    if p.is_file()
]

image_extensions = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

image_files = [
    p for p in all_files
    if p.suffix.lower() in image_extensions
]

print(
    f"Image files discovered in dataset: "
    f"{len(image_files):,}"
)

filename_lookup = defaultdict(list)

for path in image_files:
    filename_lookup[path.name].append(path)

missing_images = []
ambiguous_images = []
corrupt_images = []

for _, row in tqdm(
    images_df.iterrows(),
    total=len(images_df),
    desc="Checking images"
):

    file_name = Path(row["file_name"]).name

    matches = filename_lookup.get(
        file_name,
        []
    )

    if len(matches) == 0:
        missing_images.append(
            {
                "image_id": row["id"],
                "file_name": row["file_name"]
            }
        )

        continue

    if len(matches) > 1:
        ambiguous_images.append(
            {
                "image_id": row["id"],
                "file_name": row["file_name"],
                "matches": [
                    str(x)
                    for x in matches
                ]
            }
        )

    try:
        with Image.open(matches[0]) as img:
            img.verify()

    except Exception as e:
        corrupt_images.append(
            {
                "image_id": row["id"],
                "file_name": row["file_name"],
                "error": str(e)
            }
        )

print("\nValidation complete.")

print(
    f"Missing images   : {len(missing_images):,}"
)

print(
    f"Ambiguous files  : {len(ambiguous_images):,}"
)

print(
    f"Corrupt images   : {len(corrupt_images):,}"
)
```

```python
print("VALIDATING BOUNDING BOXES")

image_lookup = (
    images_df
    .set_index("id")
    .to_dict("index")
)

invalid_boxes = []

for _, row in tqdm(
    annotations_df.iterrows(),
    total=len(annotations_df),
    desc="Checking annotations"
):

    image_id = row["image_id"]
    bbox = row["bbox"]

    problem = None

    if image_id not in image_lookup:
        problem = "image_id_not_found"

    elif not isinstance(bbox, list) or len(bbox) != 4:
        problem = "invalid_bbox_format"

    else:

        x, y, width, height = bbox

        image_width = image_lookup[
            image_id
        ]["width"]

        image_height = image_lookup[
            image_id
        ]["height"]

        if width <= 0:
            problem = "non_positive_width"

        elif height <= 0:
            problem = "non_positive_height"

        elif x < 0:
            problem = "negative_x"

        elif y < 0:
            problem = "negative_y"

        elif x + width > image_width + 1:
            problem = "bbox_exceeds_image_width"

        elif y + height > image_height + 1:
            problem = "bbox_exceeds_image_height"

    if problem:
        invalid_boxes.append(
            {
                "annotation_id": row["id"],
                "image_id": image_id,
                "category_id": row["category_id"],
                "bbox": bbox,
                "problem": problem
            }
        )

print(
    f"Invalid annotations: "
    f"{len(invalid_boxes):,}"
)

if invalid_boxes:
    invalid_boxes_df = pd.DataFrame(
        invalid_boxes
    )

    display(
        invalid_boxes_df.head(20)
    )
else:
    print(
        "No invalid bounding boxes detected."
    )
```

```python
print("CHECKING EXACT DUPLICATE IMAGES")

referenced_paths = []

for _, row in images_df.iterrows():

    file_name = Path(
        row["file_name"]
    ).name

    matches = filename_lookup.get(
        file_name,
        []
    )

    if matches:
        referenced_paths.append(
            matches[0]
        )

hash_to_files = defaultdict(list)

for path in tqdm(
    referenced_paths,
    desc="Hashing images"
):

    try:
        sha256 = hashlib.sha256()

        with open(path, "rb") as f:

            for chunk in iter(
                lambda: f.read(1024 * 1024),
                b""
            ):
                sha256.update(chunk)

        hash_to_files[
            sha256.hexdigest()
        ].append(path)

    except Exception:
        pass

duplicate_groups = {
    h: paths
    for h, paths in hash_to_files.items()
    if len(paths) > 1
}

duplicate_file_count = sum(
    len(paths)
    for paths in duplicate_groups.values()
)

print(
    f"Duplicate groups: "
    f"{len(duplicate_groups):,}"
)

print(
    f"Images involved in duplicates: "
    f"{duplicate_file_count:,}"
)

if duplicate_groups:

    for idx, paths in enumerate(
        list(duplicate_groups.values())[:10],
        start=1
    ):

        print(f"\nDuplicate group {idx}:")

        for path in paths:
            print(
                "  ",
                path.relative_to(
                    DATASET_ROOT
                )
            )
```

```python
def resolve_image_path(file_name):
    file_name = str(file_name)

    direct_path = DATASET_ROOT / file_name

    if direct_path.exists():
        return direct_path

    base_name = Path(file_name).name

    matches = filename_lookup.get(
        base_name,
        []
    )

    if matches:
        return matches[0]

    return None


test_paths = []

for _, row in images_df.head(10).iterrows():

    resolved = resolve_image_path(
        row["file_name"]
    )

    test_paths.append(
        {
            "image_id": row["id"],
            "file_name": row["file_name"],
            "resolved_path": str(resolved)
            if resolved
            else None
        }
    )

display(
    pd.DataFrame(test_paths)
)
```

```python
def show_annotated_image(
    image_id,
    figsize=(12, 8)
):

    if image_id not in image_lookup:
        print(
            f"Image ID {image_id} not found."
        )
        return

    image_info = image_lookup[
        image_id
    ]

    image_path = resolve_image_path(
        image_info["file_name"]
    )

    if image_path is None:
        print(
            "Could not resolve image path:"
        )
        print(
            image_info["file_name"]
        )
        return

    image = Image.open(
        image_path
    ).convert("RGB")

    image_np = np.array(image)

    plt.figure(
        figsize=figsize
    )

    plt.imshow(
        image_np
    )

    ax = plt.gca()

    for annotation in annotations_by_image[
        image_id
    ]:

        x, y, width, height = (
            annotation["bbox"]
        )

        category_name = (
            CATEGORY_ID_TO_NAME.get(
                annotation["category_id"],
                "Unknown"
            )
        )

        confidence_text = ""

        if "score" in annotation:
            confidence_text = (
                f" {annotation['score']:.2f}"
            )

        rectangle = plt.Rectangle(
            (x, y),
            width,
            height,
            fill=False,
            linewidth=2
        )

        ax.add_patch(
            rectangle
        )

        ax.text(
            x,
            max(
                0,
                y - 5
            ),
            f"{category_name}{confidence_text}",
            fontsize=10,
            backgroundcolor="white"
        )

    plt.title(
        f"Image ID: {image_id}\n"
        f"{Path(image_info['file_name']).name}"
    )

    plt.axis("off")
    plt.tight_layout()
    plt.show()
```
