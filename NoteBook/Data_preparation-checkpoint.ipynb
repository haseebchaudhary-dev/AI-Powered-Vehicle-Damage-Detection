```python
DATASET_ROOT = Path(
    "dataset/CarDD_COCO"
)


ANNOTATION_FILE = (
    DATASET_ROOT
    / "annotations"
    / "instances.json"
)


OUTPUT_ROOT = Path(
    "Results/DATA PREPARATION"
)


PREP_DIR = Path(
    "Results/DATA PREPARATION/CARDD_COCO_PREPARATION"
)

PREP_DIR.mkdir(
    parents=True,
    exist_ok=True
)

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

assert abs(
    TRAIN_RATIO +
    VAL_RATIO +
    TEST_RATIO -
    1.0
) < 1e-6


RANDOM_SEED = 42

random.seed(
    RANDOM_SEED
)

np.random.seed(
    RANDOM_SEED
)

print(
    "Configuration loaded."
)

print(
    f"Dataset root : {DATASET_ROOT}"
)

print(
    f"Output root  : {OUTPUT_ROOT}"
)

print(
    f"Preparation  : {PREP_DIR}"
)
```

```python
if not DATASET_ROOT.exists():

    raise FileNotFoundError(
        f"Dataset directory not found:\n"
        f"{DATASET_ROOT}"
    )

if not ANNOTATION_FILE.exists():

    print(
        "\nConfigured annotation file was not found:"
    )

    print(
        ANNOTATION_FILE
    )

    print(
        "\nSearching automatically for COCO JSON files..."
    )

    candidates = []

    for json_file in DATASET_ROOT.rglob(
        "*.json"
    ):

        try:

            with open(
                json_file,
                "r",
                encoding="utf-8"
            ) as f:

                temp = json.load(f)

            if {
                "images",
                "annotations",
                "categories"
            }.issubset(temp.keys()):

                candidates.append(
                    json_file
                )

        except Exception:
            pass

    if not candidates:

        raise FileNotFoundError(
            "No valid COCO annotation JSON found."
        )



    candidate_info = []

    for path in candidates:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as f:

            temp = json.load(f)

        candidate_info.append(
            (
                path,
                len(temp["images"])
            )
        )

    ANNOTATION_FILE = max(
        candidate_info,
        key=lambda x: x[1]
    )[0]

print(
    "\nUsing annotation file:"
)

print(
    ANNOTATION_FILE
)
```

```python
with open(
    ANNOTATION_FILE,
    "r",
    encoding="utf-8"
) as f:

    COCO = json.load(f)

images = COCO["images"]
annotations = COCO["annotations"]
categories = COCO["categories"]

print("COCO DATASET")

print(
    f"Images      : {len(images):,}"
)

print(
    f"Annotations : {len(annotations):,}"
)

print(
    f"Categories  : {len(categories):,}"
)
```

```python

categories_sorted = sorted(
    categories,
    key=lambda x: x["id"]
)


COCO_TO_YOLO = {
    category["id"]: index
    for index, category
    in enumerate(categories_sorted)
}

YOLO_TO_NAME = {
    index: category["name"]
    for index, category
    in enumerate(categories_sorted)
}

COCO_TO_NAME = {
    category["id"]: category["name"]
    for category in categories_sorted
}

print("CATEGORY MAPPING")

for category in categories_sorted:

    coco_id = category["id"]

    yolo_id = COCO_TO_YOLO[
        coco_id
    ]

    name = category["name"]

    print(
        f"COCO ID {coco_id:>3} "
        f"→ YOLO ID {yolo_id:>2} "
        f"→ {name}"
    )
```

```python
images_df = pd.DataFrame(
    images
)

images_df = images_df[
    [
        "id",
        "file_name",
        "width",
        "height"
    ]
].copy()

images_df = images_df.rename(
    columns={
        "id": "image_id"
    }
)

print(
    images_df.head()
)
```

```python
annotations_df = pd.DataFrame(
    annotations
)

annotations_df[
    "category_name"
] = annotations_df[
    "category_id"
].map(
    COCO_TO_NAME
)

print(
    annotations_df[
        [
            "id",
            "image_id",
            "category_id",
            "category_name",
            "bbox"
        ]
    ].head()
)
```

```python
print("RESOLVING IMAGE FILES")

image_extensions = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

filename_lookup = defaultdict(list)

for path in DATASET_ROOT.rglob("*"):

    if (
        path.is_file()
        and path.suffix.lower()
        in image_extensions
    ):

        filename_lookup[
            path.name
        ].append(path)

resolved_paths = []

missing_files = []

for _, row in tqdm(
    images_df.iterrows(),
    total=len(images_df),
    desc="Resolving images"
):

    file_name = str(
        row["file_name"]
    )

    direct_path = (
        DATASET_ROOT
        / file_name
    )

    if direct_path.exists():

        resolved_paths.append(
            str(direct_path)
        )

        continue

    basename = Path(
        file_name
    ).name

    matches = filename_lookup.get(
        basename,
        []
    )

    if matches:

        resolved_paths.append(
            str(matches[0])
        )

    else:

        resolved_paths.append(
            None
        )

        missing_files.append(
            file_name
        )

images_df[
    "source_path"
] = resolved_paths

print(
    f"Resolved images: "
    f"{images_df['source_path'].notna().sum():,}"
)

print(
    f"Missing images: "
    f"{images_df['source_path'].isna().sum():,}"
)

if missing_files:

    raise RuntimeError(
        "Some COCO images could not be resolved."
    )
```

```python
image_ids = images_df[
    "image_id"
].tolist()

yolo_class_ids = list(
    range(
        len(categories_sorted)
    )
)

image_id_to_index = {
    image_id: index
    for index, image_id
    in enumerate(image_ids)
}

multi_label_matrix = np.zeros(
    (
        len(image_ids),
        len(categories_sorted)
    ),
    dtype=int
)

for _, annotation in annotations_df.iterrows():

    image_id = annotation[
        "image_id"
    ]

    category_id = annotation[
        "category_id"
    ]

    yolo_class_id = COCO_TO_YOLO[
        category_id
    ]

    image_index = image_id_to_index[
        image_id
    ]

    multi_label_matrix[
        image_index,
        yolo_class_id
    ] = 1

print("MULTI-LABEL REPRESENTATION")
print(
    "Matrix shape:",
    multi_label_matrix.shape
)

print(
    "Rows = images"
)

print(
    "Columns = damage classes"
)
```

```python
verification_rows = []

for class_id in range(
    len(categories_sorted)
):

    verification_rows.append(
        {
            "yolo_class_id": class_id,
            "class_name": YOLO_TO_NAME[
                class_id
            ],
            "images_containing_class":
                int(
                    multi_label_matrix[
                        :,
                        class_id
                    ].sum()
                )
        }
    )

verification_df = pd.DataFrame(
    verification_rows
)

print("IMAGE-LEVEL CLASS DISTRIBUTION")

display(
    verification_df
)
```

```python
classes_per_image = (
    multi_label_matrix.sum(
        axis=1
    )
)

multi_class_mask = (
    classes_per_image > 1
)

single_class_mask = (
    classes_per_image == 1
)

print("MULTI-CLASS ANALYSIS")


print(
    f"Single-class images : "
    f"{single_class_mask.sum():,}"
)

print(
    f"Multi-class images  : "
    f"{multi_class_mask.sum():,}"
)

print(
    f"Multi-class ratio   : "
    f"{multi_class_mask.mean() * 100:.2f}%"
)
```

```python


X_dummy = np.zeros(
    (
        len(images_df),
        1
    )
)

splitter_1 = (
    MultilabelStratifiedShuffleSplit(
        n_splits=1,
        test_size=TEST_RATIO,
        random_state=RANDOM_SEED
    )
)

train_val_indices, test_indices = next(
    splitter_1.split(
        X_dummy,
        multi_label_matrix
    )
)

print("FIRST SPLIT")

print(
    f"Train + Validation: "
    f"{len(train_val_indices):,}"
)

print(
    f"Test              : "
    f"{len(test_indices):,}"
)

```

```python
relative_val_ratio = (
    VAL_RATIO
    /
    (TRAIN_RATIO + VAL_RATIO)
)

splitter_2 = (
    MultilabelStratifiedShuffleSplit(
        n_splits=1,
        test_size=relative_val_ratio,
        random_state=RANDOM_SEED
    )
)

train_indices_relative, val_indices_relative = next(
    splitter_2.split(
        X_dummy[
            train_val_indices
        ],
        multi_label_matrix[
            train_val_indices
        ]
    )
)

train_indices = (
    train_val_indices[
        train_indices_relative
    ]
)

val_indices = (
    train_val_indices[
        val_indices_relative
    ]
)

print("FINAL SPLIT")

print(
    f"Train      : {len(train_indices):,}"
)

print(
    f"Validation : {len(val_indices):,}"
)

print(
    f"Test       : {len(test_indices):,}"
)

print(
    f"Total      : "
    f"{len(train_indices) + len(val_indices) + len(test_indices):,}"
)
```

```python
split_records = []

for index in train_indices:

    split_records.append(
        {
            "image_id": images_df.iloc[
                index
            ]["image_id"],
            "split": "train"
        }
    )

for index in val_indices:

    split_records.append(
        {
            "image_id": images_df.iloc[
                index
            ]["image_id"],
            "split": "val"
        }
    )

for index in test_indices:

    split_records.append(
        {
            "image_id": images_df.iloc[
                index
            ]["image_id"],
            "split": "test"
        }
    )

split_df = pd.DataFrame(
    split_records
)

split_df = split_df.merge(
    images_df,
    on="image_id",
    how="left"
)

print(
    split_df["split"]
    .value_counts()
)
```

```python
def get_split_class_distribution(
    split_name
):

    split_image_ids = set(
        split_df[
            split_df["split"]
            == split_name
        ]["image_id"]
    )

    split_annotations = (
        annotations_df[
            annotations_df[
                "image_id"
            ].isin(
                split_image_ids
            )
        ]
    )

    counts = (
        split_annotations[
            "category_name"
        ]
        .value_counts()
    )

    result = []

    for category in categories_sorted:

        name = category[
            "name"
        ]

        result.append(
            {
                "category_name": name,
                "count": int(
                    counts.get(
                        name,
                        0
                    )
                )
            }
        )

    result_df = pd.DataFrame(
        result
    )

    result_df[
        "percentage"
    ] = (
        result_df["count"]
        /
        result_df["count"].sum()
        * 100
    )

    return result_df


train_distribution = (
    get_split_class_distribution(
        "train"
    )
)

val_distribution = (
    get_split_class_distribution(
        "val"
    )
)

test_distribution = (
    get_split_class_distribution(
        "test"
    )
)

print("TRAIN CLASS DISTRIBUTION")

display(
    train_distribution
)

print("VALIDATION CLASS DISTRIBUTION")

display(
    val_distribution
)

print("TEST CLASS DISTRIBUTION")

display(
    test_distribution
)
```

```python
distribution_comparison = (
    train_distribution[
        [
            "category_name",
            "count"
        ]
    ]
    .rename(
        columns={
            "count":
                "train_count"
        }
    )
)

distribution_comparison = (
    distribution_comparison
    .merge(
        val_distribution[
            [
                "category_name",
                "count"
            ]
        ].rename(
            columns={
                "count":
                    "val_count"
            }
        ),
        on="category_name"
    )
    .merge(
        test_distribution[
            [
                "category_name",
                "count"
            ]
        ].rename(
            columns={
                "count":
                    "test_count"
            }
        ),
        on="category_name"
    )
)

distribution_comparison[
    "train_pct"
] = (
    distribution_comparison[
        "train_count"
    ]
    /
    distribution_comparison[
        "train_count"
    ].sum()
    * 100
)

distribution_comparison[
    "val_pct"
] = (
    distribution_comparison[
        "val_count"
    ]
    /
    distribution_comparison[
        "val_count"
    ].sum()
    * 100
)

distribution_comparison[
    "test_pct"
] = (
    distribution_comparison[
        "test_count"
    ]
    /
    distribution_comparison[
        "test_count"
    ].sum()
    * 100
)

print("SPLIT DISTRIBUTION COMPARISON")

display(
    distribution_comparison
)
```

```python
def coco_bbox_to_yolo(
    bbox,
    image_width,
    image_height
):
    """
    Convert COCO bbox:

        [x_min, y_min, width, height]

    into YOLO normalized format:

        x_center
        y_center
        width
        height
    """

    x_min, y_min, width, height = bbox

    x_center = (
        x_min + width / 2
    )

    y_center = (
        y_min + height / 2
    )

    x_center /= image_width
    y_center /= image_height

    width /= image_width
    height /= image_height

    return (
        x_center,
        y_center,
        width,
        height
    )
```

```python
annotations_by_image = defaultdict(
    list
)

for annotation in annotations:

    annotations_by_image[
        annotation["image_id"]
    ].append(
        annotation
    )

print(
    f"Images with annotations: "
    f"{len(annotations_by_image):,}"
)
```

```python
print("CREATING YOLO DATASET")

image_lookup = (
    images_df
    .set_index(
        "image_id"
    )
    .to_dict(
        "index"
    )
)

split_lookup = dict(
    zip(
        split_df[
            "image_id"
        ],
        split_df[
            "split"
        ]
    )
)

conversion_errors = []

for image_id, image_info in tqdm(
    image_lookup.items(),
    desc="Converting dataset"
):

    split_name = split_lookup[
        image_id
    ]

    source_path = Path(
        image_info[
            "source_path"
        ]
    )

    file_name = Path(
        image_info[
            "file_name"
        ]
    ).name

    destination_image = (
        OUTPUT_ROOT
        / "images"
        / split_name
        / file_name
    )

    destination_label = (
        OUTPUT_ROOT
        / "labels"
        / split_name
        / f"{Path(file_name).stem}.txt"
    )

    try:


        shutil.copy2(
            source_path,
            destination_image
        )


        image_width = image_info[
            "width"
        ]

        image_height = image_info[
            "height"
        ]

        yolo_lines = []

        for annotation in (
            annotations_by_image[
                image_id
            ]
        ):

            category_id = annotation[
                "category_id"
            ]

            yolo_class_id = (
                COCO_TO_YOLO[
                    category_id
                ]
            )

            bbox = annotation[
                "bbox"
            ]

            (
                x_center,
                y_center,
                width,
                height
            ) = coco_bbox_to_yolo(
                bbox,
                image_width,
                image_height
            )


            values = [
                x_center,
                y_center,
                width,
                height
            ]

            if not all(
                np.isfinite(
                    values
                )
            ):

                raise ValueError(
                    "Non-finite YOLO values."
                )

            if not (
                0 <= x_center <= 1
                and
                0 <= y_center <= 1
                and
                0 < width <= 1
                and
                0 < height <= 1
            ):

                raise ValueError(
                    f"Invalid normalized bbox: "
                    f"{values}"
                )

            yolo_lines.append(
                (
                    f"{yolo_class_id} "
                    f"{x_center:.8f} "
                    f"{y_center:.8f} "
                    f"{width:.8f} "
                    f"{height:.8f}"
                )
            )

        with open(
            destination_label,
            "w",
            encoding="utf-8"
        ) as f:

            f.write(
                "\n".join(
                    yolo_lines
                )
            )

    except Exception as e:

        conversion_errors.append(
            {
                "image_id": image_id,
                "file_name": file_name,
                "error": str(e)
            }
        )

print(
    "\nDataset conversion complete."
)

print(
    f"Conversion errors: "
    f"{len(conversion_errors):,}"
)

if conversion_errors:

    display(
        pd.DataFrame(
            conversion_errors
        ).head(20)
    )

    raise RuntimeError(
        "Dataset conversion encountered errors."
    )
```

```python
print("VALIDATING YOLO LABELS")

label_errors = []

total_yolo_annotations = 0

for split_name in [
    "train",
    "val",
    "test"
]:

    label_dir = (
        OUTPUT_ROOT
        / "labels"
        / split_name
    )

    image_dir = (
        OUTPUT_ROOT
        / "images"
        / split_name
    )

    image_files_split = [
        p for p in image_dir.iterdir()
        if p.is_file()
    ]

    for image_path in image_files_split:

        label_path = (
            label_dir
            / f"{image_path.stem}.txt"
        )

        if not label_path.exists():

            label_errors.append(
                {
                    "split": split_name,
                    "image": image_path.name,
                    "error":
                        "missing_label_file"
                }
            )

            continue

        try:

            with open(
                label_path,
                "r",
                encoding="utf-8"
            ) as f:

                lines = [
                    line.strip()
                    for line in f
                    if line.strip()
                ]

            for line_number, line in enumerate(
                lines,
                start=1
            ):

                parts = line.split()

                if len(parts) != 5:

                    label_errors.append(
                        {
                            "split": split_name,
                            "image":
                                image_path.name,
                            "line":
                                line_number,
                            "error":
                                "incorrect_field_count"
                        }
                    )

                    continue

                class_id = int(
                    parts[0]
                )

                values = [
                    float(x)
                    for x in parts[1:]
                ]

                if not (
                    0 <= class_id
                    < len(categories_sorted)
                ):

                    label_errors.append(
                        {
                            "split": split_name,
                            "image":
                                image_path.name,
                            "line":
                                line_number,
                            "error":
                                "invalid_class_id"
                        }
                    )

                if not all(
                    np.isfinite(values)
                ):

                    label_errors.append(
                        {
                            "split": split_name,
                            "image":
                                image_path.name,
                            "line":
                                line_number,
                            "error":
                                "non_finite_value"
                        }
                    )

                x_center, y_center, w, h = values

                if not (
                    0 <= x_center <= 1
                    and
                    0 <= y_center <= 1
                    and
                    0 < w <= 1
                    and
                    0 < h <= 1
                ):

                    label_errors.append(
                        {
                            "split": split_name,
                            "image":
                                image_path.name,
                            "line":
                                line_number,
                            "error":
                                "value_out_of_range"
                        }
                    )

                total_yolo_annotations += 1

        except Exception as e:

            label_errors.append(
                {
                    "split": split_name,
                    "image":
                        image_path.name,
                    "error":
                        str(e)
                }
            )

print(
    f"YOLO annotations checked: "
    f"{total_yolo_annotations:,}"
)

print(
    f"Validation errors: "
    f"{len(label_errors):,}"
)

if label_errors:

    display(
        pd.DataFrame(
            label_errors
        ).head(30)
    )

    raise RuntimeError(
        "YOLO validation failed."
    )

else:

    print(
        "All YOLO labels passed validation."
    )
```

```python
print("IMAGE / LABEL COUNT VALIDATION")

split_file_statistics = []

for split_name in [
    "train",
    "val",
    "test"
]:

    image_dir = (
        OUTPUT_ROOT
        / "images"
        / split_name
    )

    label_dir = (
        OUTPUT_ROOT
        / "labels"
        / split_name
    )

    image_count = len(
        list(
            image_dir.iterdir()
        )
    )

    label_count = len(
        list(
            label_dir.iterdir()
        )
    )

    split_file_statistics.append(
        {
            "split": split_name,
            "images": image_count,
            "labels": label_count,
            "difference":
                image_count -
                label_count
        }
    )

split_file_statistics_df = pd.DataFrame(
    split_file_statistics
)

display(
    split_file_statistics_df
)

if (
    split_file_statistics_df[
        "difference"
    ] != 0
).any():

    raise RuntimeError(
        "Image/label count mismatch detected."
    )
```
